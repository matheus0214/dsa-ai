### Exercício 6: Resolver uma Equação Quadrática

Escreva um script em Python que resolve a equação x^2 − 5x + 6 = 0 e imprime as raízes.

```python
def quadratic_equation():
    a = 1
    b = -5
    c = 4

    delta = b**2 - 4 * a * c

    if delta < 0:
        return 'delta value is equal to 0, not has any real result'
        
    x1 = (-b + np.sqrt(delta)) / 2 * a
    x2 = (-b - np.sqrt(delta)) / 2 * a

    return x1, x2

quadratic_equation()
```

### Exercício 7: Equação Quadrática com Entrada do Usuário

Modifique o programa anterior para aceitar coeficientes a, b, e c de um usuário e resolver a equação quadrática correspondente.

```python
def quadratic_equation(a, b, c):
    if a == 0:
        return "'a' value should be diff from 0"

    delta = b ** 2 - 4 * a * c

    if delta < 0:
        return 'Unable to resolve'
    elif delta == 0:
        return (-b + np.sqrt(delta)) / 2 * a
    
    return ((-b + np.sqrt(delta)) / 2 * a, (-b - np.sqrt(delta)) / 2 * a)

quadratic_equation(1, -5, 4)
    
```

### Exercício 9: Cálculo de Logaritmos

Escreva um script que calcula o logaritmo de base 10 de um número fornecido pelo usuário. Certifique-se de tratar o caso em que o logaritmo não pode ser definido (números negativos e zero).

```python
def bounds(b, x):
    m = 0

    while x > 0:
        m += 1
        x //= b

    return m-1, m

def bounds_sqrt(a):
    v = []

    for x in range(len(str(a)) + 1):
        a = np.round(np.sqrt(a), 6)
        if x != 0:
            v.append(a)

    return min(v), max(v)

def log_calc(a):
    if a < 0 or a == 0:
        return 'Unable to calculate the log'

    base = 10

    # min_b, max_b = bounds(base, a)

    min_b, max_b = bounds_sqrt(a)
    precision = 1e-3

    exp = np.round(min_b + (max_b - min_b) / 2, 5)
    guess = np.round(base ** exp, 3)
    count = 1

    while abs((guess) - a) > precision:
        if guess == a:
            break
        elif guess > a:
            max_b = exp
        else:
            min_b = exp

        exp = np.round(min_b + (max_b - min_b) / 2, 5)

        guess = np.round(base ** exp, 3)

        count +=1
    
    return count, exp

calc = 50
print(log_calc(calc))

# bounds(10, 50)
bounds_sqrt(calc) # [np.float64(1.630689), np.float64(2.659148)]
# (16, np.float64(1.698969))
# (12, np.float64(1.698974))
```
